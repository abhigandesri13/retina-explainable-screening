# Kaggle Environment Check

Run this notebook before installing or changing packages. Kaggle refreshes its notebook images regularly; capture the actual runtime rather than assuming package versions.

```python
import sys, os, platform, subprocess, importlib.metadata as metadata
from pathlib import Path

print("Python:", sys.version)
print("Platform:", platform.platform())
print("Working directory:", Path.cwd())
print("Kaggle input mounted:", Path("/kaggle/input").exists())
print("Kaggle input entries:", [p.name for p in Path("/kaggle/input").iterdir()] if Path("/kaggle/input").exists() else "not mounted")
```

```python
packages = ["torch", "torchvision", "timm", "albumentations", "cv2", "numpy", "pandas", "sklearn", "PIL", "mlflow", "optuna"]
for package in packages:
    try:
        module = __import__(package)
        version = getattr(module, "__version__", "version not exposed")
        print(f"{package:18s} {version}")
    except Exception as exc:
        print(f"{package:18s} NOT AVAILABLE ({type(exc).__name__}: {exc})")
```

```python
import torch
print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("Torch CUDA build:", torch.version.cuda)
if torch.cuda.is_available():
    print("GPU count:", torch.cuda.device_count())
    for index in range(torch.cuda.device_count()):
        print(f"GPU {index}:", torch.cuda.get_device_name(index))
```
